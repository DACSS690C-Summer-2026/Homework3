## Homework 2 - Optimization

Option A: Vitamin Supplements Problem

```python
import pulp as pp
import pandas as pd
```

### Initialize the model

In the code below I created the model, giving it the name 'supplements-problem' and setting the sense to minimize.

```python
# set up the model
model = pp.LpProblem(name='supplements-problem',
                     sense=pp.LpMinimize) # minimize
```

### Declare the variables

The vitamin supplements problem has two variables: SuperVit and NewHealth, both of which are set below. I set them both to integer values because it wouldn't make sense to buy a portion of a pill.

```python
# supervit variable
SuperVit = pp.LpVariable(name="SuperVit",
                    lowBound=0,
                    cat='Integer')

# newhealth variable
NewHealth = pp.LpVariable(name="NewHealth",
                 lowBound=0,
                 cat='Integer')
```

### Function to optimize using given coefficients

The problem gives the following coefficients:

* SuperVit: \$0.20
* NewHealth: \$0.30

I assigned these coefficients below and set up the objective function of:

$ SuperVitCoeff * SuperVit + NewHealthCoeff * NewHealth $


```python
# cost of each SuperVit tab
SuperVitCoeff=.20
# cost of each NewHealth tab
NewHealthCoeff=.30

# objective function
obj_func = SuperVitCoeff * SuperVit + NewHealthCoeff * NewHealth
```

### Represent the constraints

The problem gives the following daily requirements for vitamins (the constraints):

* 60 mill - vitamin C
* 1000 mill - calcium
* 18 mill - iron
* 20 mill - niacin
* 360 mill - magnesium

These constraints are set up in the code below as well as the amount of each vitamin in a single SuperVit and NewHealth pill.

```python
# SUBJECT TO:
C1= pp.LpConstraint(name='VitC',
                    e= 20*SuperVit + 30*NewHealth,
                    rhs = 60,
                    sense = pp.LpConstraintGE)
C2 = pp.LpConstraint(name = 'Calc',
                     e = 500*SuperVit + 250*NewHealth,
                     rhs = 1000,
                     sense = pp.LpConstraintGE)
C3 = pp.LpConstraint(name = 'Iron',
                     e = 9*SuperVit + 2*NewHealth,
                     rhs = 18,
                     sense = pp.LpConstraintGE)
C4 = pp.LpConstraint(name = 'Niacin',
                     e = 2*SuperVit + 10*NewHealth,
                     rhs = 20,
                     sense = pp.LpConstraintGE)
C5 = pp.LpConstraint(name = 'Magn',
                     e = 60*SuperVit + 90*NewHealth,
                     rhs = 360,
                     sense = pp.LpConstraintGE)
```

### Build model

The objective function and all of the constraints above are added to the final model.

```python
model += obj_func
model += C1
model += C2
model += C3
model += C4
model += C5
```

### Solve the model

The model is solved below. The output is 1 meaning the model was successfully solved.

```python
# model.solve()
model.solve(pp.PULP_CBC_CMD(msg=0))
```

### Model report

The model report is printed below.

The report shows that the minimal amount of spending per day on vitamins would be $1.20 by purchasing 2 NewHealth and 3 SuperVit.

```python
"Model Status",pp.LpStatus[model.status]
Optimal_Decision={"Optimal Solution - Minimal Spending":pp.value(model.objective)}
Optimal_Decision.update({v.name: v.varValue for v in model.variables()})
Optimal_Decision


pd.DataFrame.from_dict(Optimal_Decision,orient='index',columns=['info']).map('{:,.2f}'.format)
```
